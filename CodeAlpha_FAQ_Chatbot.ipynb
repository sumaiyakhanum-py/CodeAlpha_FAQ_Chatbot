```python
import re
import ipywidgets as widgets
from IPython.display import display, HTML
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. FAQ Dataset (Questions and Answers)
faq_knowledge_base = {
    "what is ml": "Machine learning is a subset of AI that enables systems to learn and improve from experience without being explicitly programmed.",
    "what is python": "Python is a high-level, interpreted programming language known for its simplicity and readability.",
    "what is a chatbot": "A chatbot is a software application used to conduct an online chat conversation via text or text-to-speech.",
    "what is nlp": "Natural Language Processing (NLP) is a branch of AI that helps computers understand, interpret, and manipulate human language.",
    "what is sql": "SQL stands for Structured Query Language, used for managing and manipulating relational databases.",
    "what is data science": "Data Science combines domain expertise, programming skills, and math/stat knowledge to extract meaningful insights from data.",
    "what is computer vision": "Computer Vision is a field of AI that enables computers to derive meaningful information from digital images and videos.",
    "what is generative ai": "Generative AI refers to algorithms that can be used to create new content, including audio, code, images, text, and videos.",
    "how can i learn ai": "You can start learning AI by mastering Python, understanding basic linear algebra and statistics, and taking courses on platforms like Coursera, NPTEL, or YouTube."
}

questions = list(faq_knowledge_base.keys())
answers = list(faq_knowledge_base.values())

# 2. NLP Preprocessing function
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text.strip()

cleaned_questions = [preprocess_text(q) for q in questions]

# 3. TF-IDF Vectorizer and Fit
vectorizer = TfidfVectorizer(stop_words='english')
question_vectors = vectorizer.fit_transform(cleaned_questions)

# 4. Smart Answer Matching Function
def get_smart_answer(user_question):
    cleaned_user = preprocess_text(user_question)
    if not cleaned_user:
        return "Please ask a question!"

    # Check for simple greetings
    greetings = ["hi", "hello", "hey", "good morning", "good evening"]
    if cleaned_user in greetings:
        return "Hello! 👋 I'm your Smart AI FAQ Assistant. How can I help you today?"

    user_vector = vectorizer.transform([cleaned_user])
    similarity_scores = cosine_similarity(user_vector, question_vectors)[0]

    best_match_index = similarity_scores.argmax()
    best_score = similarity_scores[best_match_index]

    # Threshold check for best matching answer
    if best_score < 0.2:
        return "Sorry, I don't have an answer for that right now. Try asking about Python, ML, NLP, Data Science, or SQL!"

    return answers[best_match_index]

# 5. Interactive UI Chat Widget
chat_output = widgets.Output()

user_input = widgets.Text(
    placeholder='Type your question here...',
    description='You:',
    layout=widgets.Layout(width='70%')
)

send_button = widgets.Button(
    description='Send',
    button_style='primary'
)

def send_message(button):
    question = user_input.value.strip()
    if question == "":
        return

    response = get_smart_answer(question)

    with chat_output:
        print(f"You: {question}")
        print(f"Chatbot: {response}\n" + "-"*40)

    user_input.value = ""

send_button.on_click(send_message)

display(HTML("<h3>🤖 Smart AI FAQ Assistant</h3>"))
display(chat_output)
display(widgets.HBox([user_input, send_button]))

with chat_output:
    print("Chatbot: Hello! 👋 Ask me anything about AI, Python, ML, NLP, Data Science, or SQL!\n" + "-"*40)
```
