```python
import requests

url = "https://api.mymemory.translated.net/get"

params = {
    "q": "Hello, how are you?",
    "langpair": "en|fr"
}

response = requests.get(url, params=params)

data = response.json()

print("Original: Hello, how are you?")
print("Translated:", data["responseData"]["translatedText"])
```

```python
import requests

languages = {
    "English": "en",
    "French": "fr",
    "Spanish": "es",
    "Hindi": "hi",
    "Telugu": "te",
    "Urdu": "ur"
}

print("Available languages:")
for language in languages:
    print("-", language)

source = input("Enter source language: ")
target = input("Enter target language: ")
text = input("Enter text: ")

source_code = languages[source]
target_code = languages[target]

url = "https://api.mymemory.translated.net/get"

params = {
    "q": text,
    "langpair": f"{source_code}|{target_code}"
}

response = requests.get(url, params=params)
data = response.json()

print("\nOriginal:", text)
print("Translated:", data["responseData"]["translatedText"])
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, Javascript
import requests

languages = {
    "English": "en",
    "French": "fr",
    "Spanish": "es",
    "Hindi": "hi",
    "Telugu": "te"
}

source = widgets.Dropdown(
    options=list(languages.keys()),
    description="From:"
)

target = widgets.Dropdown(
    options=list(languages.keys()),
    description="To:"
)

text_box = widgets.Textarea(
    placeholder="Enter text here...",
    description="Text:",
    layout=widgets.Layout(width="500px", height="100px")
)

translate_button = widgets.Button(description="Translate")
copy_button = widgets.Button(description="Copy Translation")

output = widgets.Output()
translated_text = ""

def translate_text(b):
    global translated_text

    with output:
        clear_output()

        text = text_box.value.strip()

        if not text:
            print("Please enter some text.")
            return

        url = "https://api.mymemory.translated.net/get"

        params = {
            "q": text,
            "langpair": f"{languages[source.value]}|{languages[target.value]}"
        }

        response = requests.get(url, params=params)
        data = response.json()

        translated_text = data["responseData"]["translatedText"]

        print("Translated Text:")
        print(translated_text)

def copy_translation(b):
    if translated_text:
        display(Javascript(
            f"navigator.clipboard.writeText({translated_text!r})"
        ))
        print("Translation copied!")
    else:
        print("Please translate something first.")

translate_button.on_click(translate_text)
copy_button.on_click(copy_translation)

display(source, target, text_box, translate_button, copy_button, output)
```
